```python
import os

from dotenv import load_dotenv
load_dotenv()

from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-4o-mini", temperature=0)


response = llm.invoke('Hola soy luis y mi telefono es 5566666666')
response.text()
```

```python
from pydantic import BaseModel, Field

class ContactInfo(BaseModel):
    name: str = Field(description="The name of the person")
    email: str = Field(description="The email address of the person")
    phone: str = Field(description="The phone number of the person")
    tone: int = Field(description="The tone of the person",ge=0, le=100)
    age: int = Field(description="The age of the person" )
    sentiment: str = Field(description="The sentiment conversation of the person")

llm_with_structured_output = llm.with_structured_output(schema=ContactInfo)
messages = [
    ('system',"You are a helpful assist that can extract contact information from a given conversation."),#System Prompt
    ("user",'Hola soy luis y mi telefono es 5566666666')
]
response = llm_with_structured_output.invoke(messages)
#response.pretty_print() #Error
response.name
```
