```python
from langchain_openai import ChatOpenAI

gpt4o_llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
response = gpt4o_llm.invoke("Hola ¿cómo estás?")
response.text
```

```python
gpt3o_llm = ChatOpenAI(model="gpt-3o-mini", temperature=0)
response = gpt4o_llm.invoke("Hola ¿cómo estás?")
response.text
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage

msg_1 = SystemMessage(content="How can I help you?")
msg_2 = HumanMessage(content="Me llamo Juan")
msg_3 = AIMessage(content="Hola Juan cómo estas?")
msg_4 = HumanMessage(content="Cómo me llamo?")
history = [msg_1,msg_2,msg_3,msg_4]
```

```python
history = [msg_1,msg_2,msg_3,msg_4]
response = gpt3o_llm.invoke(history) #Se manda el historial y lo recordo
response.pretty_print()
```

```python
from langchain_anthropic import ChatAnthropic

claude_llm = ChatAnthropic(model="claude-3-5-sonnet-20240620", temperature=0)
response = claude_llm.invoke(history)
response.pretty_print()
```
